# MATHEMATICAL FUNCTIONS

**1+1=2** is a statement about the Mathematical Function "ADDITION":

- if the numbers 1 and 1 are provided as **input** to the **Mathematical Function** "ADDITION", we get the **output** 2 


Some other examples of Mathematical Functions:

  - 1,2,3,4,5,6 → **ADDITION** → 21    (several inputs, 1 output)<br>
  - 1 → **ADD_ONE** → 2   (1 input, 1 output)<br>
  - 2,6,10 → **MULTIPLY** →  120     <br>
  - 2,6,10 → **AVERAGE** → 6 <br>
  - -12 ,  7,  11,  18  →  **SHOW_SMALLEST_AND_LARGEST** →  -12, 18  (Note we can have functions with **multiple outputs**)
  - -12, 7, 11, 17  →  **HOW_MANY_NEGATIVE_AND_HOW_MANY_POSITIVE**  →  1, 3

All of these Mathematical Functions:
- take zero or more **number(s)** as input,
- produce one or more **numbers(s)** as output
- are **deterministic** - meaning they always give the same answer for the same input.   1+1 is always equals to 2...<br> 


What about these:








- These might be horrendously complex MATHEMATICAL FUNCTIONS, but **THEY ARE STILL MATHEMATICAL FUNCTIONS** :
    - they take numbers as input, 
    - they produce numbers as output
    - they are deterministic

# NEURAL NETWORKS

Like a program, Neural Networks:

- also take numbers as input and produce numbers as output.
- are deterministic (they provide the same answer when given the same input)







But unlike a program, the Neural Network:

- gives **APPROXIMATE** answers, typically as "probabilities/likelihoods"
- rather than being "programmed", it is "trained" using thousands/millions of input-output pairs
- Neural networks let us **approximate the behaviour of massively complex mathematical functions**
- Think of a Neural Network as a **Universal Mathematical Function Approximator**  (More @ https://www.youtube.com/watch?v=xg4bIeJTVF0)

## Our task from 20000 feet: 
Neural Networks are made up of **Perceptrons** - often hundreds, sometime thousands or millions of them..

### A Perceptron:



A neural network will have:
  - an input layer, which is simply **the input** numbers
  - 0 or more "hidden" layers, **containing perceptrons**
  - an output layer containing **the answer**
  - **adjustable weights** connecting everything

Here is an untrained neural network for recognising MNIST characters. 

The blue parts are our perceptrons<br>

The challenge is to **adjust the weigths** to get the behaviour we want..


### Why Perceptrons? What makes them so useful?

Perceptrons can represent simple mathematical functions exactly:

Consider the function: **y = mx+c**

This can be replicated by the Perceptron:



```python
def f(x): return 5*x+2
plot_function(f, 'x', '5x+2')
```

But that is not very exciting or useful..

Let's make a small change to this perceptron:



We call the extra block the **"activation function"** (because it determines how much the perceptron is "activated")

We can also swap C and 1 and rename C to "bias" to arrive at a perceptron:




```python
def rectified_linear(weight,bias,x):
  y=weight*x+bias
  return torch.clip(y, 0.)

@interact(weight=1.5, bias=1.5)
def plot_relu(weight, bias):
    plot_function( partial(rectified_linear, weight, bias))
```

This is much more useful, because **you can theoretically combine a lot of these (perhaps millions) to closely approximate any x,y function**..




### Consider (just) two perceptrons:

```python
def double_relu(w1, b1, w2, b2,x):
   return rectified_linear(w1,b1,x)+  rectified_linear(w2,b2,x)

@interact(w1=-1.5, b1=-1.5, w2=1.5, b2=1.5)
def plot_double_relu(w1, b1, w2, b2):
    plot_function( partial(double_relu, w1, b1,w2,b2))
```

With enough perceptrons and tweaking of the weights, you could replicate any x-y mathematical function.



And that is with just a very simple NN: one input ( X ) and one output ( Y ) :)


A perceptron usually has many more x-weight pairs too:




A typical NN with a hidden layer and many inputs, can simulate hugely complex "multi-dimensional" mathematical functions - ones that are good enough to recognise hand-written digits.. 


### A better activation function
It turns out that max(a,0) is good but a different **activation function** is even more useful as:

- it still allows us to create any function by adding more and more perceptrons
- it again disallows negative values, but it also squashes the output to between 0 and 1
- it prevents explosions in values across the Neural Network,
- it is called the Sigmoid Function  (more @ https://towardsdatascience.com/derivative-of-the-sigmoid-function-536880cf918e)




```python
def sigmoid(x): return 1/(1+torch.exp(-x))
plot_function(torch.sigmoid, title='Activation Function "Sig"', min=-4, max=4)
```

This gives us a full Perceptron:






Our challenge is to find the weights in the network such that the Neural Network best approximates our hypothetical perfect mathematical function.

Interesting Thought! ->  If we had endless time, we could simply adjust the weights randomly until we found an excellent fit, but we want an efficient way to do this...


## Our task from 5000 feet


### Training a Neural Network

- Let's call one snapshot of the NN at work **an iteration**, where:
    -  we present an image (in this example a picture of the digit 9) and 
    -  we see the output and 
    -  we adjust the weights a little to reduce the "error" (also called "cost")



#### The Cost..

For any iteration, the "Cost" or "Badness" of the NN is measured using the "Mean Squared Error" (more @ https://en.wikipedia.org/wiki/Mean_squared_error)

- 0 = perfect, and
- the larger the Cost, the worse the results..

We want the overall cost, to be as small as possible

- Each weight in the NN has an effect on the Cost (Imagine setting a weight to infinity, it will clearly effect the Cost..)
- Increasing or decreasing each weight (will alter the outputs and therefore) will alter the current Cost
- At the end of each iteration we want to **NUDGE** each weight a little bit, to reduce the current Cost
- **But how do we know in which direction to nudge each weight?**

Ideally:
- It would be nice to know the function relating each weight to the cost...
- For each weight (say w5) we could theoretically...
  - "freeze the NN", then vary w5 through a wide range of values, and plot w5 against cost
  
  - we could then simply update w5 to the location where cost is a minimum
  - then repeat for the other ~100000 weights.
  - but this is WAY WAY too expensive

Realistically:
- Instead we do something MUCH MUCH CHEAPER. For each weight..
  - we find the current **slope** of our weight against cost using the "Chain Rule"
  - For example w5 against cost


  - we do not see the dashed line (the function between w5 and cost)
  - we DO see the gradient
  - we nudge the weight "downhill"
  - this is called gradient descent





#### Slopes/Derivatives... A Mathematical diversion...

- If **y=f(x)**, then the **DERIVATIVE of y w.r.t x**, (written **dy/dx**), gives the **SLOPE** of the function.
- For example if **y = pow(x,2)**, then the **DERIVATIVE of y w.r.t x**, (written **dy/dx**), is equal to **2x**, and describes the **SLOPE** of the function
- Calculating the slope (taking the derivate) of a simple function is VERY EASY and CHEAP

```python
def f(x): return pow(x,2)
plot_function(f, 'x', 'y = pow(x,2)')

def dydx(x): return 2*x

plot_function(dydx, 'x', 'dydx = 2x')
```


#### The Chain Rule
- To determine the slopes, we use the **Chain Rule**, which makes use of 2 important facts..
    - For each iteration, all relationships and values in our neural network are clearly defined..

    - Given a large function, we can calculate **the slope of its input to its output, by chaining together the individual slopes within it**


- Which gives us the slopes of each weight to cost for the current iteration ..


- We can then NUDGE each weight downhill...
- More @ https://mattmazur.com/2015/03/17/a-step-by-step-backpropagation-example and https://www.youtube.com/watch?v=hM74RH82LyI

# HOW A NEURAL NETWORK LEARNS... Ground level..

We now have all we need to train a neural network...

1) Initialize the network by selecting random values between 0 and 1 for all the weights (and biases)
2) For each image:
   - Plug in the input values of an image, to **set all the values in the NN** ("**Feed Forwards**")
   - Calculate the **slopes** of each weight w.r.t the cost, by using the chain rule.. ("**Back propogation**")
   - Nudge the weights downhill
   - Repeat from step 2 for all images in our data set, to make **one Epoch**

Perform a few Epochs and we are done!
